# Student Performance Analysis
**EDA Project | Python**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
math = pd.read_csv("student-mat.csv", sep=";")
por = pd.read_csv("student-por.csv", sep=";")
math.head()
```

## 1. Basic information

```python
print("Math shape:", math.shape)
print("Portuguese shape:", por.shape)
math.info()
math.describe()
```

## 2. Data cleaning

```python
print("Math missing values:", math.isnull().sum().sum())
print("Portuguese missing values:", por.isnull().sum().sum())
print("Math duplicates:", math.duplicated().sum())
print("Portuguese duplicates:", por.duplicated().sum())
```

## 3. Final grade analysis

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(math["G3"], bins=15)
ax[0].set_title("Math final grades")
ax[1].hist(por["G3"], bins=15)
ax[1].set_title("Portuguese final grades")
plt.show()
print("Average Math G3:", round(math.G3.mean(), 2))
print("Average Portuguese G3:", round(por.G3.mean(), 2))
```

## 4. Study time and grades

```python
sns.boxplot(data=math, x="studytime", y="G3")
plt.title("Math grade by study time")
plt.show()
sns.boxplot(data=por, x="studytime", y="G3")
plt.title("Portuguese grade by study time")
plt.show()
print(math.groupby("studytime").G3.mean().round(2))
```

## 5. Past failures

```python
sns.boxplot(data=math, x="failures", y="G3")
plt.title("Math grade by past failures")
plt.show()
print(math.groupby("failures").G3.mean().round(2))
```

## 6. Absences and grades

```python
plt.scatter(math.absences, math.G3)
plt.xlabel("Absences")
plt.ylabel("Final grade")
plt.title("Absences vs Math final grade")
plt.show()
print("Correlation:", round(math[["absences", "G3"]].corr().iloc[0,1], 2))
```

## 7. Grade progression

```python
plt.scatter(por.G1, por.G3)
plt.xlabel("First period grade")
plt.ylabel("Final grade")
plt.title("Portuguese G1 vs G3")
plt.show()
print("Correlation:", round(por[["G1", "G3"]].corr().iloc[0,1], 2))
```

## 8. Correlation heatmap

```python
plt.figure(figsize=(11, 8))
sns.heatmap(math.select_dtypes(include=np.number).corr(), cmap="coolwarm", center=0)
plt.title("Math correlation heatmap")
plt.show()
```

## 9. Merge the two datasets

```python
merge_cols = ['school', 'sex', 'age', 'address', 'famsize', 'Pstatus', 'Medu', 'Fedu', 'Mjob', 'Fjob', 'reason', 'nursery', 'internet']
merged = math.merge(por, on=merge_cols, suffixes=("_math", "_por"))
print("Matched students:", len(merged))
merged.head()
```

## 10. Findings

```python
print("Math pass share (G3 >= 10):", round((math.G3 >= 10).mean()*100, 1), "%")
print("Portuguese pass share (G3 >= 10):", round((por.G3 >= 10).mean()*100, 1), "%")
print("Wanting higher education:", round((math.higher == "yes").mean()*100, 1), "%")
```

### Conclusion
- Study time and past failures show useful patterns with final grades.
- Previous grades are useful for understanding later performance.
- Absences can be checked along with other student factors.
- Schools can provide extra support to students who are struggling.

*These patterns do not prove cause and effect.*
